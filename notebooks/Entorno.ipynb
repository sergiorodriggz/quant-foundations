```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd             
```

```python
# NumPy: crear y calcular con un array
datos = np.array([1, 2, 3])
print("Media:", np.mean(datos))

# Pandas: crear y mostrar un DataFrame
df = pd.DataFrame({"x": datos, "y": datos ** 2})
display(df)

# Matplotlib: graficar los datos
df.plot(x="x", y="y", marker="o", title="Demostración")
plt.show()

```
